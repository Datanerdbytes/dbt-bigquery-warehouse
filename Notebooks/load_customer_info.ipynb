```python
import sys
from sqlalchemy import create_engine, text

# Connection Parameters
SERVER = 'localhost' 
DATABASE = 'Demo_Database'
USERNAME = 'sa'
PASSWORD = 'YourStrongPassword123!'
DRIVER = 'ODBC Driver 18 for SQL Server'


connection_string = (
    f"mssql+pyodbc://{USERNAME}:{PASSWORD}@{SERVER}/{DATABASE}?"
    f"driver={DRIVER}&TrustServerCertificate=yes"
)

try:
    engine = create_engine(connection_string, pool_pre_ping=True)
    
    with engine.connect() as connection:
        result = connection.execute(text("SELECT DB_NAME() AS CurrentDB, SUSER_SNAME() AS CurrentUser;"))
        row = result.fetchone()
        print(f"Successfully connected to database: '{row.CurrentDB}' as user: '{row.CurrentUser}'")

except Exception as e:
    print("Connection failed!")
    print(f"Error details:\n{e}")
    sys.exit(1)
```

```python
engine = create_engine(connection_string, fast_executemany=True)
df = pd.read_csv(r'/Users/roelsomido/Dataset/sql-data-warehouse-project/datasets/source_crm/cust_info.csv')
df.columns = df.columns.str.strip()

TARGET_SCHEMA = 'bronze'
TARGET_TABLE = 'crm_cust_info'

try:
    df.to_sql (
        name=TARGET_TABLE,
        con=engine,
        schema=TARGET_SCHEMA,
        if_exists='append',
        index=False,
        chunksize=10000
    )
    print(f"Successfully appended {len(df)} rows into {TARGET_SCHEMA}.{TARGET_TABLE}")
except Exception as e:
    print(f"Failed to append data: {e}")
```
